```python
#Data cleaning
# Create copies of the original datasets
matches_clean = matches.copy()
deliveries_clean = deliveries.copy()

print("Backup copies created successfully!")
```

```python
print("Missing values in MATCHES:")
print(matches_clean.isnull().sum())

print("\nMissing values in DELIVERIES:")
print(deliveries_clean.isnull().sum())
```

```python
# Fill missing values in categorical columns

deliveries_clean['extra_type'] = deliveries_clean['extra_type'].fillna('none')

deliveries_clean['dismissal_kind'] = deliveries_clean['dismissal_kind'].fillna('not_out')

deliveries_clean['fielder'] = deliveries_clean['fielder'].fillna('none')

print("Missing categorical values filled successfully!")
```

```python
print("Missing values after cleaning:")
print(deliveries_clean.isnull().sum())
```

```python
# Total runs
runs_per_season = (
    deliveries_clean
    .groupby('match_id')['total_runs']
    .sum()
    .reset_index()
    .merge(
        matches_clean[['match_id', 'season']],
        on='match_id'
    )
    .groupby('season')['total_runs']
    .sum()
)

print(runs_per_season)
```

```python
# Checking whether the toss winner also won the match

matches_clean['toss_match_winner'] = (
    matches_clean['toss_winner'] == matches_clean['winner']
)

print(matches_clean['toss_match_winner'].value_counts())
```

```python
# Calculate the percentage
toss_win_percentage = (
    matches_clean['toss_match_winner'].mean() * 100
)

print("Percentage of matches where toss winner also won:",
      round(toss_win_percentage, 2), "%")
```

```python
toss_result = matches_clean['toss_match_winner'].value_counts()

toss_result.index = [
    'Toss Winner Lost Match',
    'Toss Winner Won Match'
]

plt.figure(figsize=(8, 5))

plt.bar(
    toss_result.index,
    toss_result.values,
    color='purple'
)

plt.title('Toss Winner vs Match Winner')
plt.xlabel('Result')
plt.ylabel('Number of Matches')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
# Calculate total wickets by season

wickets_per_season = (
    deliveries_clean
    .merge(
        matches_clean[['match_id', 'season']],
        on='match_id',
        how='left'
    )
    .groupby('season')['is_wicket']
    .sum()
)

print("Total wickets by season:")
print(wickets_per_season)
```

```python
plt.figure(figsize=(10, 5))

wickets_per_season.plot(
    kind='bar',
    color='teal'
)

plt.title('Total IPL Wickets by Season (2018–2026)')
plt.xlabel('Season')
plt.ylabel('Total Wickets')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
most_wicket_season = wickets_per_season.idxmax()
most_wickets = wickets_per_season.max()

print("Season with most wickets:", most_wicket_season)
print("Total wickets:", most_wickets)
```

```python
# Create a list of all teams from team1 and team2

team1_matches = matches_clean[['match_id', 'team1']].rename(
    columns={'team1': 'team'}
)

team2_matches = matches_clean[['match_id', 'team2']].rename(
    columns={'team2': 'team'}
)

all_team_matches = pd.concat([team1_matches, team2_matches])

matches_played = (
    all_team_matches
    .groupby('team')['match_id']
    .nunique()
    .sort_values(ascending=False)
)

print("Matches played by each team:")
print(matches_played)
```

```python
# Find wins by each team

team_wins = (
    matches_clean[matches_clean['winner'] != 'Tie']
    ['winner']
    .value_counts()
)

print("Wins by each team:")
print(team_wins)
```

```python
team_performance = pd.DataFrame({
    'Matches Played': matches_played,
    'Wins': team_wins
})

team_performance['Wins'] = team_performance['Wins'].fillna(0)

team_performance['Win Percentage'] = (
    team_performance['Wins'] /
    team_performance['Matches Played'] * 100
)

team_performance = team_performance.sort_values(
    'Win Percentage',
    ascending=False
)

print(team_performance.round(2))
```

```python
plt.figure(figsize=(12, 6))

team_performance['Win Percentage'].sort_values(
    ascending=False
).plot(
    kind='bar',
    color='teal'
)

plt.title('IPL Team Win Percentage (2018–2026)')
plt.xlabel('Team')
plt.ylabel('Win Percentage (%)')
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
best_team = team_performance['Win Percentage'].idxmax()
best_percentage = team_performance['Win Percentage'].max()

print("Best team by win percentage:", best_team)
print("Win percentage:", round(best_percentage, 2), "%")
```

```python
#Find the most-used venues
venue_matches = (
    matches_clean
    .groupby('venue')['match_id']
    .nunique()
    .sort_values(ascending=False)
)

print("Top 10 venues by number of matches:")
print(venue_matches.head(10))
```

```python
top_10_venues = venue_matches.head(10).sort_values()

plt.figure(figsize=(10, 6))

plt.barh(
    top_10_venues.index,
    top_10_venues.values,
    color='slateblue'
)

plt.title('Top 10 IPL Venues by Number of Matches (2018–2026)')
plt.xlabel('Number of Matches')
plt.ylabel('Venue')

plt.tight_layout()
plt.show()
```

```python
#Find the highest-scoring venue
venue_runs = (
    deliveries_clean
    .merge(
        matches_clean[['match_id', 'season', 'venue']],
        on='match_id',
        how='left'
    )
    .groupby(['venue', 'match_id'])['total_runs']
    .sum()
    .reset_index()
)

average_runs_venue = (
    venue_runs
    .groupby('venue')['total_runs']
    .mean()
    .sort_values(ascending=False)
)

print("Top 10 venues by average runs per match:")
print(average_runs_venue.head(10).round(2))
```

```python
top_scoring_venues = average_runs_venue.head(10).sort_values()

plt.figure(figsize=(10, 6))

plt.barh(
    top_scoring_venues.index,
    top_scoring_venues.values,
    color='seagreen'
)

plt.title('Top 10 Venues by Average Runs per Match (2018–2026)')
plt.xlabel('Average Runs per Match')
plt.ylabel('Venue')

plt.tight_layout()
plt.show()
```

```python
toss_success_percentage = (
    toss_won_match /
    (toss_won_match + toss_lost_match)
) * 100

print(
    "Toss winner match-win percentage:",
    round(toss_success_percentage, 2),
    "%"
)
```

```python
labels = ['Toss Winner Won Match', 'Toss Winner Lost Match']
values = [toss_won_match, toss_lost_match]

plt.figure(figsize=(8, 5))

plt.bar(
    labels,
    values,
    color='mediumpurple'
)

plt.title('Toss Winner vs Match Winner (2018–2026)')
plt.xlabel('Result')
plt.ylabel('Number of Matches')

plt.tight_layout()
plt.show()
```
